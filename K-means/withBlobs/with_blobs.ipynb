##### Playing with BLOBS

```python
import pandas as pd
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
```

```python
X, y = make_blobs(n_samples = 400, n_features = 2, random_state = 23, centers = 3, )

plt.figure(figsize = (6,6))
plt.scatter(X[:,0], X[:,1] ,s = 50, marker = '.', edgecolor = 'black', alpha = 0.8)
```

```python
kmeans = KMeans(n_clusters = 3, random_state = 23)
kmeans.fit(X)

labels = kmeans.labels_
centers = kmeans.cluster_centers_

plt.figure(figsize = (6,6))
plt.scatter(X[:,0], X[:,1], c=labels, cmap = 'viridis', s = 40, edgecolor = 'black', )
plt.scatter(centers[:,0], centers[:,1], c='red', s = 100, edgecolor = 'black', marker = 'X')
```

```python
import numpy as np

to_pred = np.array([[1.34, 3], [123,234]])
pred = kmeans.predict(to_pred)

print(pred)

mask = labels == 1

plt.scatter(to_pred[:,0], to_pred[:,1])
plt.scatter(X[mask,0], X[mask,1], c = 'red', edgecolor = 'black', s = 10, alpha = 0.4)
```
